# Exploratory Data Analysis (EDA)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\sumit\OneDrive\Desktop\Own Data Anlyst Project\Retail_Business_Intelligence\Dataset\cleaned_data.csv")
df["Order_Date"] = pd.to_datetime(df["Order Date"])
```

```python
monthly_sales = df.groupby(df["Order_Date"].dt.to_period("M"))["Sales"].sum()
monthly_sales.plot(figsize=(12,5), marker="o")
plt.title("Monthly Sales")
plt.ylabel("Sales")
plt.show()
```

```python
monthly_profit = df.groupby(df["Order_Date"].dt.to_period("M"))["Profit"].sum()
monthly_profit.plot(figsize=(12,5), marker="o")
plt.title("Monthly Profit")
plt.ylabel("Profit")
plt.show()
```

```python
category_sales = df.groupby("Category of Goods")["Sales"].sum()

category_sales.plot(kind="bar", figsize=(8,5))
plt.title("Category Sales")
plt.show()
```

```python
state_sales = df.groupby("State")["Sales"].sum().sort_values(ascending=False).head(10)

state_sales.plot(kind="bar")
plt.title("Top 10 States by Sales")
plt.show()
```

```python
customer_orders = df.groupby("Customer Name")["Order ID"].count()

customer_orders.hist(bins=30)
plt.title("Customer Distribution")
plt.show()
```

```python
top_products = df.groupby("Product Name")["Sales"].sum().nlargest(10)

top_products.plot(kind="bar")
plt.title("Top Products")
plt.show()
```

```python
plt.scatter(df["Discount"], df["Profit"])
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.title("Discount vs Profit")
plt.show()
```

# Customer Segmentation (RFM + K-Means)

```python
import pandas as pd

snapshot_date = df["Order Date"].max() + pd.Timedelta(days=1)

rfm = df.groupby("Customer ID").agg({
    "Order Date": lambda x: (snapshot_date - x.max()).days,
    "Order ID": "nunique",
    "Sales": "sum"
})

rfm.columns = ["Recency","Frequency","Monetary"]
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(rfm)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=4, random_state=42)

rfm["Cluster"] = kmeans.fit_predict(rfm_scaled)
```

```python
rfm.groupby("Cluster")[["Recency","Frequency","Monetary"]].mean()
```

```python
cluster_map = {
    0:"VIP",
    1:"Regular",
    2:"New",
    3:"At Risk"
}

rfm["Customer_Type"] = rfm["Cluster"].map(cluster_map)
```

# Forecasting

```python
from sklearn.linear_model import LinearRegression
import numpy as np

daily = df.groupby("Order_Date")["Sales"].sum().reset_index()

daily["Day"] = np.arange(len(daily))

model = LinearRegression()

model.fit(daily[["Day"]], daily["Sales"])

daily["Predicted"] = model.predict(daily[["Day"]])
```

```python
plt.figure(figsize=(12,5))

plt.plot(daily["Order_Date"], daily["Sales"], label="Actual")
plt.plot(daily["Order_Date"], daily["Predicted"], label="Predicted")

plt.legend()
plt.show()
```

```python
from statsmodels.tsa.arima.model import ARIMA

model = ARIMA(daily["Sales"], order=(1,1,1))

result = model.fit()

forecast = result.forecast(30)
```
